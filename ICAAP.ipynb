```python
df1 = pd.read_excel(r"C:\Users\Usuario\Desktop\Estudos-Estag\ICAAP\Base_Full_Auditoria.xlsx")
df2 = pd.read_excel(r"C:\Users\Usuario\Desktop\Estudos-Estag\ICAAP\Base_Unica.xlsx")
```

```python
#filtrando apenas por ICAAP
df1 = df1[df1["Auditoria"] == 'ICAAP']
```

```python
df1
```

```python
#Tirando nulos

df1 = df1.dropna(subset=["Código auditoria"])
df2 = df2.dropna(subset=["Descrição"])

#Tirando duplicados
df1 = df1.drop_duplicates(subset=["Código auditoria"])
```

```python
#Concatenando colunas 
df2['Descrição + Código'] = df2['Código auditoria'] + ' - ' + df2['Descrição']
```

```python
df2['Descrição + Código']
```

```python
df2.columns
```

```python
df_join = df1.merge(
    df2,
    on='Código auditoria',
    how='inner'  # ou 'left' se quiser manter tudo do df1
)
```

```python
df_join.columns
```

```python
df_final = df_join[
    [
        'Risco',
        'PMAPA_x',
        'Código auditoria',
        'Descrição + Código',
        'Nível',
        'Status',
    ]
]
```

```python
df_final['Status'] = df_final['Status'].str.strip().str.lower()

df_final = df_final[df_final['Status'].isin(['aberto', 'em andamento'])]
```

```python
df_final.columns
```

```python
df_final['Ano'] = 2025
```
